```python
import pandas as pd
```

```python
# Load the CSV file
df = pd.read_csv("Penalty Kick Dataset_2020–2025.csv", sep=";")
```

```python
list(df)
```

```python
df = df.drop(columns=['del1', 'del2', 'del3', 'del4'])
```

```python
# 3. How many Penalties are there?
total_penalties = len(df)
print(f"Total Penalties: {total_penalties}")
```

```python
# 4. Penalties not during regular time
# Regular time = Minutes 1 to 90 (+ stoppage time). Extra time = 91-120+. Shootouts = >120 or special markers.
# Ensure 'Minute' is numeric
df["Minute_clean"] = pd.to_numeric(df["Minute"], errors="coerce")

# Outside standard 90 mins (Extra time or Shootouts)
outside_90_mins = df[df["Minute_clean"] > 90]
shootouts_or_et = df[df["Minute_clean"] > 120]
```

```python
print(f"\nPenalties taken after minute 90 (Extra Time / Stoppage / Shootouts): {len(outside_90_mins)}")
print(f"Penalties taken after minute 120 (Shootout indicators): {len(shootouts_or_et)}")
```

```python
# 5. Penalties per Player (Shooters)
# Fallback to player_id if player_name has missing values
player_col = "player_name" if "player_name" in df.columns else "player_id"
penalties_per_player = df[player_col].value_counts()

print(f"\nTotal Unique Shooters: {df[player_col].nunique()}")
print("\nTop 10 Penalty Takers:")
print(penalties_per_player.head(10))
```

```python
# 6. Penalties per Keeper
# Note the column spelling in this schema: 'goalkeer_name' or 'goalkeeper_id'
keeper_col = "goalkeer_name" if "goalkeer_name" in df.columns else "goalkeeper_id"
penalties_per_keeper = df[keeper_col].value_counts()

print(f"\nTotal Unique Goalkeepers Faced: {df[keeper_col].nunique()}")
print("\nTop 10 Goalkeepers Facing Penalties:")
print(penalties_per_keeper.head(10))
```

```python
print("\n--- Penalties by Zone ---")
zone_breakdown = df["Zone"].value_counts()
print(zone_breakdown)
```

```python
# Goalie Side vs Kicker Side matrix
print("\nKicker Side vs Goalie Side (Counts):")
print(pd.crosstab(df["Kicker_Side"], df["Goalie_Side"], margins=True))
```
