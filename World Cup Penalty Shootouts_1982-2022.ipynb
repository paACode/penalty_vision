```python
import pandas as pd
```

```python
# Load the CSV file
df = pd.read_csv("WorldCupShootouts.csv")
```

```python
list(df)
```

```python
# 1. How many Matches are there?
total_matches = df["Game_id"].nunique()
print(f"Total Matches (Shootouts): {total_matches}")
```

```python
# 2. How many Teams are there?
unique_teams = df["Team"].unique()
print(f"Total Teams: {len(unique_teams)}")
print(f"Teams: {', '.join(sorted(unique_teams))}")
```

```python
# 3. How many Penalties are there?
total_penalties = len(df)
print(f"Total Penalties: {total_penalties}")
```

```python
# 4. How many were NOT during regular time?
# Every penalty in this dataset is from a penalty shootout (100% outside regular time)
penalties_outside_regular = total_penalties
print(f"Penalties outside regular time: {penalties_outside_regular} (100%)")
```

```python
# 5 & 6. Penalties per Player / Keeper Note
# The CSV tracks dive directions ('L', 'C', 'R') and footedness rather than player/keeper names.
# Instead, we break down penalties taken/faced per Team and Keeper Dive Directions:

print("\n--- Shootout Penalties Taken Per Team ---")
print(df["Team"].value_counts())
```

```python
print("\n--- Goalkeeper Dive Directions ---")
keeper_dives = df["Keeper"].value_counts().rename(index={"L": "Left", "R": "Right", "C": "Center"})
print(keeper_dives)
```

```python
print("\n--- Penalties by Zone ---")
zone_breakdown = df["Zone"].value_counts()
print(zone_breakdown)
```
