```python
from pathlib import Path
import re
import pandas as pd
```

```python
output_csv = Path("../../data/processed/penalty_plays.csv")
penalties = pd.read_csv(output_csv)
```

```python
output_csv_others = Path("../../data/processed_others/penalty_plays.csv")
penalties_others = pd.read_csv(output_csv_others)
```

```python
penalties.columns
```

```python
counts = penalties["penalty_direction"].value_counts()
percents = (
    penalties["penalty_direction"]
    .value_counts(normalize=True)
    .mul(100)
    .round(1)
    .astype(str)
    + "%"
)

# Combine into a single DataFrame and sort alphabetically by penalty direction
penalty_directions = (
    pd.concat([counts, percents], axis=1, keys=["count", "percentage"])
    .sort_index()
)

penalty_directions.head(20)
```

```python
import pandas as pd

outcome_cols = ["goal", "save", "miss"]

# 1. Total counts per direction
counts = penalties.groupby("penalty_direction").size().rename("total_count")
percents = (
    (counts / counts.sum() * 100)
    .round(1)
    .astype(str)
    + "%"
).rename("total_pct")

# 2. Outcomes breakdown per direction
outcomes = penalties.groupby("penalty_direction")[outcome_cols].sum()

# Rate of each outcome within each direction
outcome_pcts = (
    outcomes.div(counts, axis=0)
    .mul(100)
    .round(1)
    .astype(str)
    + "%"
).add_suffix("_pct")

summary_df = pd.concat([counts, percents, outcomes, outcome_pcts], axis=1).sort_index()
summary_df.head(20)
```
