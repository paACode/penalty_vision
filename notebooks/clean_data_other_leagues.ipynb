```python
from pathlib import Path
import pandas as pd
folder_path = {"commentary_data": "commentary", "keyEvents_data": "keyEvents", "plays_data": "plays"}
years = ["2024", "2025"]

merged_data = {}

data_dir = Path("../data/processed_others")

data_dir.mkdir(parents=True, exist_ok=True)

print("Starting to merge CSV files...")

for folder, prefix in folder_path.items():
    dfs =[]
    for year in years:
        for file_name in file_names_others:
            file_path = f"../data/raw/{folder}/{prefix}_{year}_{file_name}.csv"
            print(f"Checking file: {file_path}")
            if Path(file_path).exists():
                print(f"Reading file...")
                df = pd.read_csv(file_path)
                df["year"] = year
                df["competition"] = file_name
                dfs.append(df)
                print(f"Loaded: {file_path}")
            else:
                print(f"File does not exist: {file_path}")
        # Merge all CSVs from this folder
    if dfs:
        merged_data[folder] = pd.concat(dfs, ignore_index=True)

print("Merging completed. Saving merged files...")
merged_data["commentary_data"].to_csv("../data/processed_others/merged_commentary.csv", index=False)
merged_data["keyEvents_data"].to_csv("../data/processed_others/merged_keyEvents.csv", index=False)
merged_data["plays_data"].to_csv("../data/processed_others/merged_plays.csv", index=False)





```
